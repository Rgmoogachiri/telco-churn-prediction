# 03 - Cleaning
Goal: produce a clean dataset (data/processed/telco_clean.csv) from the raw file.

```python
import pandas as pd

df = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")
df.shape
```

```python
# Convert TotalCharges to numeric
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
df["TotalCharges"].isna().sum()
```

```python
df.isna().sum()
```

fill those 11 with 0, after proving they're new customers

```python
assert df.loc[df["TotalCharges"].isna(), "tenure"].eq(0).all()
df["TotalCharges"] = df["TotalCharges"].fillna(0)
df["TotalCharges"].isna().sum()
```

```python
df["customerID"].is_unique
```

```python
df = df.drop(columns="customerID")
```

The ID is unique for every customer, so it carries no pattern for the model to learn.

```python
df.head()
```

Turn the target into 0 and 1. No = 0 (stayed) and Yes = 1 (Left)

```python
df["Churn"] = df["Churn"].map({"No": 0, "Yes": 1})
df["Churn"].value_counts()
```

```python
df.shape
```

```python
df.isna().sum().sum()
```

```python
df.dtypes
```

```python
df.to_csv("../data/processed/telco_clean.csv", index=False)
```
