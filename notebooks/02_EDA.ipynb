```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
# Temporary conversion for plotting only. The formal cleaning happens in notebook 03.
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
df["ChurnFlag"] = (df["Churn"] == "Yes").astype(int)
```

```python
df["ChurnFlag"].mean()*100
```

Overall churn rate is 26.5% (1,869 of 7,043 customers). This is the baseline to compare segments against.

```python
df.groupby("Contract")["ChurnFlag"].mean().sort_values()
```

Month-to-month customers churn at 42.7%, versus 11.3% for one-year and 2.8% for two-year contracts.

```python
df["Contract"].value_counts(normalize=True)
```

55% of customers are on month-to-month contracts. Combined with their high churn rate, they account for roughly 89% of all churners.

```python
df.groupby("InternetService")["ChurnFlag"].mean().sort_values()
```

Fiber optic customers churn at 41.9%, more than double DSL (19.0%). Customers with no internet service churn least (7.4%).

```python
df.groupby("PaymentMethod")["ChurnFlag"].mean().sort_values()
```

Electronic check customers churn at 45.3%, about three times the rate of customers paying by automatic card or bank transfer.

```python
df.groupby(["Contract", "PaymentMethod"])["ChurnFlag"].mean().unstack()
```

Electronic check has the highest churn within every contract type, so payment method carries signal beyond contract type.

```python
df.groupby(["Contract", "PaymentMethod"]).size().unstack()
```

Group sizes: month-to-month + electronic check is the largest segment (1,850 customers, ~54% churn),
accounting for over half of all churners. Two-year cells have very few churners (3 to ~20),
so those rates are indicative only.

# 02 - Exploratory data analysis

## Numeric features vs churn.

```python
sns.boxplot(data=df, x="Churn", y="tenure")
plt.show()
```

Churned customers have much shorter tenure (median ~10 months vs ~38 for retained customers). Churn is concentrated among newer customers.

```python
sns.boxplot(data=df, x="Churn", y="MonthlyCharges")
plt.show()
```

Churned customers pay more per month (median ~$80 vs ~$64). Low-bill customers (likely phone-only) rarely churn. This overlaps with fiber optic and contract type, so it is not a standalone cause.

## Summary of findings

* Overall churn is 26.5%
* Month-to-month contract is the strongest factor (42.7% churn, about 89% of churners)
* Fiber optic and electronic check customers churn the most
* Churners are newer and pay higher monthly charges
* Factors overlap, so the model needs to be checked for what really matters
